```python
from epsilon_greedy import EpsilonGreedy
from thompson_sampling import ThompsonSampling
from visualizations import Visualization
from functions import save_results
import pandas as pd
```

```python
arms = [1, 2, 3, 4]
```

```python
eg_algo = EpsilonGreedy(arms)
ts_algo = ThompsonSampling(arms)
```

```python
df_eg = eg_algo.experiment(20000)
df_ts = ts_algo.experiment(20000)
```

```python
eg_algo.report()
ts_algo.report()
```

```python
viz = Visualization(df_eg, df_ts)
viz.plot_learning_curve()
viz.plot_cumulative()
```

```python
df_all = pd.concat([df_eg, df_ts], axis=0)
save_results(df_all)
```
